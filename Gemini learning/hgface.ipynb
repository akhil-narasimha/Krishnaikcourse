```python
from sentence_transformers import SentenceTransformer
import numpy as np
from numpy.linalg import norm

model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
sentences = [
    "I sat by the river bank to watch the water.",
    "I deposited my money in the bank account.",
    "I need to withdraw cash from the financial institution."
]

# 3. Generate the dense embeddings
print("Generating embeddings...")
embeddings = model.encode(sentences)

print(f"\nShape of the embeddings matrix: {embeddings.shape}")

# 4. Your Cosine Similarity Function
def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2) / (norm(vec1) * norm(vec2))

# 5. The Moment of Truth: Let's test the math
sim_river_vs_money = cosine_similarity(embeddings[0], embeddings[1])
sim_money_vs_finance = cosine_similarity(embeddings[1], embeddings[2])

print("\n--- The Results ---")
print(f"Similarity (River Bank vs Money Bank): {sim_river_vs_money:.4f}")
print(f"Similarity (Money Bank vs Financial Inst.): {sim_money_vs_finance:.4f}")
```

```python
# Your "Database" of documents
faq_database = [
    "To reset your password, click the 'Forgot Password' link on the login screen.",
    "Our refund policy allows returns within 30 days of the original purchase date.",
    "You can update your billing address in the 'Account Settings' dashboard."
]

# A messy user search query
user_query = "How do I get my money back for this thing?"

faq_questions = model.encode(faq_database)
user_query_embeddings = model.encode([user_query])
```

```python
scores = []
for i in range(len(faq_database)):
    sim = cosine_similarity(user_query_embeddings[0], faq_questions[i])
    scores.append(sim)
    print(f"Similarity between user query and FAQ {i+1}: {sim:.4f}")
```

```python
import numpy as np
best_match_index = np.argmax(scores)
print(f"\nBest matching FAQ: {faq_database[best_match_index]} (Similarity: {scores[best_match_index]:.4f})")
```
